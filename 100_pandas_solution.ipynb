```python
## 🔹 Section 1: Basics & Series (Q1–Q20)
### Q1. Create a Pandas Series from a list of integers.
# > 💡 **Hint:** Use `pd.Series([...])`. The default index starts from 0.
import pandas as pd
import numpy as np
arr = np.array([1,2,3,4,5])
series = pd.Series(arr)
print(series)
```

```python
### Q2. Create a Series from a dictionary with custom index.
# > 💡 **Hint:** Pass a `dict` to `pd.Series()`. Keys become the index labels automatically.
dict = {"name":"Raj" , "Age":34 , "address":"pune"}
s = pd.Series(dict)
print(s)
```

```python
### Q3. Access elements using index labels.
# > 💡 **Hint:** Use `s["label"]` for a single value, `s[["a","b"]]` for multiple, and `s["a":"c"]` for a range. Label slices are **inclusive** on both ends.
s = pd.Series(['raj' , 'pallavi' ,'priya'],index=['a' , 'b' ,'c'])
print(s["a"])
print(s[["a" , "c"]])

#label slicing(inclusive)
print(s["a" : "c"])

```

```python
### Q4. Perform arithmetic operations on a Series.
# > 💡 **Hint:** Use `+`, `-`, `*`, `/`, `//`, `%`, `**` directly on the Series. Operations are **element-wise** and **index-aligned**.
s = pd.Series([45,56,70],index = ['a','b','c'])

print(s+10)
print(s*2)

#Arithmetic between two Series (index alignment)
s2 = pd.Series([5,10,15],index=['a','b','c'])
print(s+s2)

#Misaligned index
s3 = pd.Series([1,2,3],index=['a' ,'b','d'])
print("s+s3:\n",s+s3)
```

```python
### Q5. Find mean, min, max of a Series.
# > 💡 **Hint:** Use `.mean()`, `.min()`, `.max()`, `.sum()`, `.median()`, `.std()` directly on the Series.
s = pd.Series([1,2,3],index= ['a','b','c'])
print(s)
print("mean:",s.mean())
print("min:",s.min())
print("max:",s.max())
print("sum:",s.sum())
print("median:",s.median())
```

```python
### Q6. Apply cnditional filtering on a Series.
# > 💡 **Hint:** Use a boolean mask: `s[s > 50]`. You can also chain conditions with `&` and `|`.
import pandas as pd 
s = pd.Series([45,56,70],index=['a','b','c'])

result = s[s > 50]
print(result)

#multiple conditions (and)
result1= s[(s > 50) & (s<70)]
print(result1)

result2 = s[(s > 50) | (s<70)]
print(result2)

```

```python
### Q7. Replace values in a Series.
# > 💡 **Hint:** Use `.replace(old_val, new_val)` or pass a dictionary `{old: new}` to replace multiple values at once.
s = pd.Series([45,56,70],index=['a','b','c'])
replace = s.replace(56 ,90)
print(replace)

#replace multiple values
rep_mul = s.replace({45:90 , 70:100})
print(rep_mul)
```

```python
### Q56. Detect outliers in a column.
# > 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.
df = pd.DataFrame({
    "age":[10,12,14,15,18,100]
})

#step1:calculate Q1 and Q3 
Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)

IQR = Q3 - Q1 

lower_bound = Q1 - 1.5 *IQR 
upper_bound = Q3 + 1.5* IQR

outliers = df[(df["age"]< lower_bound) | (df['age'] > upper_bound)]
print(outliers)
```

```python
### Q100. Create an end-to-end analysis workflow.
import pandas as pd 
df = pd.DataFrame({
     "Department": ["IT", "IT", "HR", "HR", "Finance", "Finance"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Salary": [50000, 60000, 40000, 45000, 70000, 75000],
    "Experience": [2, 3, 5, 4, 6, 7]
})
df.to_csv("data.csv")
```

```python
#Data cleaning
df = df.dropna() 

df = df.drop_duplicates()

df['Salary']=df["Salary"].astype(int)
```

```python
#EDA
print(df.describe())

#Group analysis
print(df.groupby("Department")["Salary"].mean())
```

```python
#Feature Engineering
df["Salary_After_Bonus"] = df["Salary"]*1.10
print(df)
```

```python
#Data Transformation
df["Department"] = df["Department"].astype("category").cat.codes

#scaling (basic)
df["Salary"] = (df["Salary"] - df["Salary"].mean())/df["Salary"].std()
```

```python
#Summary Report
report = pd.pivot_table(
    df,
    values='Salary',
    index = "Department",
    columns="Gender",
    aggfunc="mean"
)
print(report)
```
